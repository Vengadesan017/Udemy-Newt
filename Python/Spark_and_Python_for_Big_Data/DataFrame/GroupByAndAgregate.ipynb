# group by and aggregate

```python
from pyspark.sql import SparkSession
```

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("GroupBY").getOrCreate()
```

```python
df = spark.read.csv('appl_stock.csv',inferSchema=True,header=True)
```

```python
df.printSchema()
```

```python
df.groupBy("Low").count().show()
```

```python
df.groupBy("Low").mean().show()
```

# min max in each grp

```python
df.groupBy("Low").min().select('Low').show()
```

```python
df.groupBy("Low").max().select('Low').show()
```

# over all min , max

```python
df.agg({'open':'max'}).show()
```

# Distinct 

```python
from pyspark.sql.functions import countDistinct, avg,stddev
df.select(countDistinct("low")).show()
```

# format the long float


```python
from pyspark.sql.functions import format_number
```

```python
sales_std = df.select(stddev("open").alias('open_data'))
```

```python
sales_std.show()
```

```python
sales_std.select(format_number('open_data',1)).show()
```

# Order by

```python
df.orderBy("open").show()
```

```python
df.orderBy(df["open"].desc()).show()
```
