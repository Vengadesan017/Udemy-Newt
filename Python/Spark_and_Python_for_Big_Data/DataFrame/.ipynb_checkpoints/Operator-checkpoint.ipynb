```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Operations").getOrCreate()
```

```python
df = spark.read.csv('appl_stock.csv',inferSchema=True,header=True)
```

```python
df
```

```python
df.printSchema()
```

```python
df.filter("Close<500").show()
```

```python
df.filter("Close<100").select('Open').show()
```

```python
# Using SQL with .select()
df.filter("Close<100").select(['Open','Close']).show()
```

```python
df.filter(df["Close"] < 100).show()
```

```python
df.filter( (df["Close"] < 100) & (df['Open'] > 100) ).show()
```

```python
df.filter( "Close < 100" & "Open > 100" ).show()
```

```python
df.filter( (df["Close"] < 100) & ~(df['Open'] > 100) ).show()
```

```python
df.filter(df["Low"] == 197.16).show()
```

```python
df.filter(df["Low"] == 197.16).collect()  # rpw so use .asDict() to iterable
```
