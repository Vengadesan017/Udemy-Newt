# Most used words (non-articles)

```python
import re
```

```python
with open("miracle_in_the_andes.txt","r", encoding="utf-8") as file:
    book = file.read()
    pattern = re.compile("[a-zA-Z]+")
    result = re.findall(pattern,book.lower())
    for i in result[:10]:
        print(i)
    print(len(result))
```

```python
d = {}
for x in result:
    if x in d.keys():
        d[x] = d[x] + 1
    else:
        d[x] = 1
```

```python

list1 = [(value, key) for (key, value) in d.items()]
sorted(list1, reverse=True)
len(list1)
```

```python

import nltk

from nltk.corpus import stopwords
english_stopwords = stopwords.words("english")
```

```python
len(english_stopwords)

```

```python
english_stopwords[:10]
```

```python
filtered_words = []
for count, word in list1:
    if word not in english_stopwords:
        filtered_words.append((word, count))
```

```python
len(filtered_words)
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer
nltk.download('vader_lexicon')
# nltk.sentiment.SentimentIntensityAnalyzer
```

```python
analyzer = SentimentIntensityAnalyzer()
```

```python

scores = analyzer.polarity_scores("Haii i am going to school, and i not take leave unwantedly")
```

```python
scores
```

```python

scores = analyzer.polarity_scores("Haii i am not going to school, and i take leave unwantedly")
```
