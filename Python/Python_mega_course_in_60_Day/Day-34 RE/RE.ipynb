```python
with open("miracle_in_the_andes.txt","r", encoding="utf-8") as file:
    book = file.read()
    pattern = re.compile(" [a-zA-z ]* love [a-zA-z ]")
    result = re.findall(pattern,book)
    for i in result[:10]:
        print(i)
    print(len(result))
```

```python
with open("miracle_in_the_andes.txt","r", encoding="utf-8") as file:
    book = file.read()
    pattern = re.compile(" [a-zA-z ,:]* love [a-zA-z ]")
    result = re.findall(pattern,book)
    for i in result[:10]:
        print(i)
    print(len(result))
```

```python
with open("miracle_in_the_andes.txt","r", encoding="utf-8") as file:
    book = file.read()
    pattern = re.compile("[a-zA-Z]+")
    result = re.findall(pattern,book.lower())
    for i in result[:10]:
        print(i)
    print(len(result))
```
